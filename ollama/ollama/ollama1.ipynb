```python
import base64
import ollama
image_path="ChatGPT Image Aug 14, 2026, 07_04_59 PM.png"
with open(image_path,"rb") as f:
    image_bytes=f.read()
image_64=base64.b64encode(image_bytes).decode("utf-8") #encoding the image in base64 bits
response=ollama.generate(model="gemma3:4b",images=[image_64],prompt="give caption to the image")
print(response.response)
```

```python
#if we want to give multiple images then we can give in the form of list 
import base64
images_path=["ChatGPT Image Aug 14, 2026, 07_04_59 PM.png","WhatsApp Image 2026-06-12 at 10.42.12 AM.jpeg"]
images_base64=[]
for i in images_path:
    with open(i,"rb") as f:
        image_read=f.read()
        images_base64.append(base64.b64encode(image_read).decode("utf-8"))
response=ollama.generate(model="gemma3:4b",images=images_base64,prompt="describe both the images")
response.response
        
```

```python
#giving the system instruction means telling the modle how to behave
response=ollama.generate(model="gemma3:4b",prompt="why does the moon glow?",system="you are a funny assistant and explain the things in funny way")
response.response
```

```python
#option parameter ke andar ham few parameters chnage kar sakte hai 
response=ollama.generate(model="gemma3:4b",prompt="why is the ocean blue",
                         options={
                             "temperature":0.3,
                             "top_p":0.5,
                             "top_k":45})
print(response.response)
```
