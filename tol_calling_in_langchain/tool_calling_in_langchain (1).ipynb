##currency converter

```python
#tool creation
from langchain_core.tools import InjectedToolArg
from typing import Annotated
@tool
def get_conversion_factor(base_currency:str,target_currency:str)->float:
  """this function fetches the currency conversion factor between a base currency and target currency"""
  url=f"https://v6.exchangerate-api.com/v6/2f90b31db65/pair/{base_currency}/{target_currency}"
  response=requests.get(url)
  return response.json()

@tool
def convert(base_currency_value:int,conversion_rate:Annotated[float,InjectedToolArg])->float:
  """given a currency conversion rate this function calculates the target currency value  from a given base currency value"""
  return base_currency_value*conversion_rate
```

```python
convert.args
```

```python
get_conversion_factor.invoke({'base_currency':'USD','target_currency':'INR'})
```

```python
convert.invoke({'base_currency_value':10,'conversion_rate':95.6696})
```

```python
llm
```

```python
llm_with_tools=llm.bind_tools([get_conversion_factor,convert])
```

```python
llm_with_tools
```

```python
messages=[HumanMessage('What is the conversion factor between usd and inr, and can u convert 10 usd into inr ')]
```

```python
messages
```

```python
ai_message=llm_with_tools.invoke(messages)
```

```python
messages.append(ai_message)
```

```python
tool_msg = get_conversion_factor.invoke(ai_message.tool_calls[0])
messages.append(tool_msg)
```

```python
ai_message = llm_with_tools.invoke(messages)
```

```python
ai_message.tool_calls
```

```python
import json

conversion_rate = None

for tool_calls in ai_message.tool_calls:

    if tool_calls['name'] == 'get_conversion_factor':
        tool_result1 = get_conversion_factor.invoke(tool_calls)
        print(tool_result1)

        conversion_rate = json.loads(tool_result1.content)['conversion_rate']
        messages.append(tool_result1)

    elif tool_calls['name'] == 'convert':
        if conversion_rate is None:
            # Get the conversion rate first
            rate_tool_call = {
                'name': 'get_conversion_factor',
                'args': {
                    'base_currency': 'USD',
                    'target_currency': 'INR'
                },
                'id': 'temp',
                'type': 'tool_call'
            }

            tool_result1 = get_conversion_factor.invoke(rate_tool_call)
            conversion_rate = json.loads(tool_result1.content)['conversion_rate']
            messages.append(tool_result1)

        tool_calls['args']['conversion_rate'] = conversion_rate

        tool_message2 = convert.invoke(tool_calls)
        messages.append(tool_message2)
```

```python
llm_with_tools.invoke(messages).content
```
