```python
from langchain.agents import create_agent
from langchain_classic import hub #isme predefined kuch prompts hote hai
```

```python
agent = create_agent(
    model=llm,
    tools=[search_tool],
    system_prompt="You are a helpful assistant. Use the search tool whenever current information is required."
)
```

```python
# Ask the agent
response = agent.invoke({
    "messages": [
        {
            "role": "user",
            "content": "what is the population of the capital of france"
        }
    ]
})

print(response["messages"][-1].content)
```
